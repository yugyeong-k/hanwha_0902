### 6. PromptTemplate

```python
# 질문 정의
template = "{country}의 수도는 어디인가요?"

# 정의한 질문을 PromptTemplate 객체로 생성
prompt_template = PromptTemplate.from_template(template)

# country에 값을 넣어 실제 프롬프트 생성

# format() : 템플릿의 변수를 채워 문자열(str)로 반환
# 단순히 완성된 문장을 만들고 싶을 때 사용
prompt1 = prompt_template.format(country="대한민국")

# invoke() : 템플릿의 변수를 채워 PromptValue 객체로 반환
# LangChain의 다음 단계(LLM, Chain 등)에 연결할 때 사용
prompt2 = prompt_template.invoke({"country": "대한민국"})

print(prompt1, type(prompt1))

# PromptValue = LangChain이 프롬프트로 처리하기 좋게 만든 객체
print(prompt2, type(prompt2))
```

```python
prompt = PromptTemplate.from_template("{topic} 에 대해 쉽게 설명해주세요.")
input = {"topic": "인공지능 모델의 학습 원리"}
```

### 7. 출력 파서 설정

```python
from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()
```

### 8. LCEL로 연결

```python
# LCEL로 연결
# chain = prompt | model
chain = prompt | model | output_parser

chain.invoke(input)
```

- 테디노트 스트리밍

```python
# 스트리밍 출력을 위한 요청
answer = chain.stream(input)

# TeddyNote의 스트리밍 출력 함수
stream_response(answer)
```

- 공식 LangChain 스트리밍

```python
# 스트리밍 방식으로 Chain 실행
answer = chain.stream(input)

# 스트리밍 결과를 하나씩 출력
for chunk in answer:
    print(chunk, end="")
    
# for chunk in chain.stream(input):
#     print(chunk, end="")
```
