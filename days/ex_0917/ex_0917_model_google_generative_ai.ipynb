```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash")

answer = llm.stream("자연어처리에 대해서 간략히 설명해 줘")

for chunk in answer:
    print(chunk.text, end="", flush=True)
```

```python
from langchain_core.prompts import PromptTemplate

model = ChatGoogleGenerativeAI(model = "gemini-3.5-flash")

prompt = PromptTemplate.from_template("예/아니오 질문에 대답하세요. {question}는 과일입니까?")

chain = prompt | model

for chunk in chain.stream({"question": "사과"}):
    print(chunk.text, end="", flush=True)
```

```python
from langchain_google_genai import HarmBlockThreshold, HarmCategory

llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-pro",
    # 안정성 지정
    safety_settings={
        # BLOCK_NONE : 항상 허용, BLOCK_ONLY_HIGH 유해 판정이 높을때만 차단
        # 성적으로 노골적인 콘텐츠
        HarmCategory.HARM_CATEGORY_SEXUALLY_EXPLICIT: HarmBlockThreshold.BLOCK_NONE,
        # 혐오 발언 
        HarmCategory.HARM_CATEGORY_HATE_SPEECH: HarmBlockThreshold.BLOCK_NONE,
        # 괴롭힘/모욕 콘텐츠
        HarmCategory.HARM_CATEGORY_HARASSMENT: HarmBlockThreshold.BLOCK_NONE,
        # 위험한 콘텐츠
        HarmCategory.HARM_CATEGORY_DANGEROUS_CONTENT: HarmBlockThreshold.BLOCK_NONE,
    },
)
```

```python
llm = ChatGoogleGenerativeAI(model = "gemini-3.5-flash")

results = llm.batch(
    [
        "대한민국 수도는?",
        "대한민국 주요 관광지 5곳 나열",
    ]
)

for chunk in results:
    print(chunk.content)
```

```python
from langchain_google_genai import GoogleGenerativeAI
from langchain.messages import HumanMessage, SystemMessage
import base64

model = ChatGoogleGenerativeAI(model= "gemini-3.5-flash", temperature= 0)

system_prompt = ("당신은 시인입니다. 당신의 임무는 주어진 이미지를 가지고 시를 작성하는 것입니다.")

user_prompt = "다음의 이미지에 대한 시를 작성해주세요."

with open("images/jeju-beach.jpg", "rb") as f:
    image_data = base64.b64encode(f.read()).decode("utf-8")

message = [
    SystemMessage(content = system_prompt),
    
    HumanMessage(
        content = [
            {
                "type" : "text",
                "text" : user_prompt,
            },
            {
                "type": "image_url",
                "image_url" : f"data:image/jpeg;base64,{image_data}",
            }   
        ]
    )
]

response = model.invoke(message)
print(response.content)
```
