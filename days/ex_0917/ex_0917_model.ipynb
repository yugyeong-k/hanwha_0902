```python
from langchain_openai import ChatOpenAI

gpt = ChatOpenAI(temperature=0, model="gpt-4o-mini")

answer = gpt.invoke("사랑이 뭔가요?")

print(answer.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI

# Chat Model
chat = ChatOpenAI()

chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "이 시스템은 여행 전문가입니다."),
    ("user", "{user_input}"),
])

chain = chat_prompt | chat
chat_result = chain.invoke({"user_input": "안녕하세요? 한국의 대표적인 관광지 3군데를 추천해주세요."})
print(chat_result)
```

```python
from langchain.chat_models import init_chat_model

# 통합 모델
model = init_chat_model("gpt-4.1")

# 모델 호출
response = model.invoke("안녕하세요, 한국의 수도는 어디인가요?")
print(response.content)
```

```python
from langchain.chat_models import init_chat_model

configurable_model = init_chat_model(
    configurable_fields=("model", "model_provider"))

# OpenAI
response_gpt = configurable_model.invoke(
    "당신의 이름은 무엇인가요?",
    config={
        "configurable": {
            "model": "gpt-4o-mini",
            "model_provider": "openai"
        }
    }
)

# Gemini
response_gemini = configurable_model.invoke(
    "당신의 이름은 무엇인가요?",
    config={
        "configurable": {
            "model": "gemini-3.5-flash",
            "model_provider": "google_genai"
        }
    }
)

print(f"GPT: {response_gpt.content}")
print(f"Gemini: {response_gemini.content}")
```

```python
from langchain.chat_models import init_chat_model

# 첫 번째 모델 (요약용)
summary_model = init_chat_model(
    model="gpt-4.1-mini",
    model_provider="openai",
    configurable_fields=(
        "model",
        "model_provider",
        "temperature",
        "max_tokens",
    ),
    config_prefix="summary",
)

# 두 번째 모델 (창작용)
creative_model = init_chat_model(
    model="gemini-3.5-flash",
    model_provider="google_genai",
    configurable_fields=(
        "model",
        "model_provider",
    ),
    config_prefix="creative",
)

# 런타임에 summary_model을 Gemini로 변경
response = summary_model.invoke(
    "다음 텍스트를 요약해주세요: "
    "랭체인은 대규모 언어 모델을 활용한 애플리케이션을 "
    "쉽게 개발할 수 있도록 도와주는 프레임워크입니다.",
    config={
        "configurable": {
            "summary_model": "gemini-3.5-flash",
            "summary_model_provider": "google_genai",
            "summary_max_tokens": 500,
        }
    },
)

print(response.content)
```
