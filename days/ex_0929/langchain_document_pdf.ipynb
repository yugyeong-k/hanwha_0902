```python
def show_metadata(docs):
    if docs:
        print("[metadata]")
        print(list(docs[0].metadata.keys()))
        print("\n[examples]")

        max_key_length = max(
            len(k) for k in docs[0].metadata.keys()
        )

        for k, v in docs[0].metadata.items():
            print(f"{k:<{max_key_length}} : {v}")
```

```python
from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(model="text-embedding-3-small")

text = "인공지능 기술이 빠르게 발전하고 있다."

vector = embeddings.embed_query(text)

print(vector[:10])
print("벡터 차원:", len(vector))
```

```python
from langchain_community.vectorstores import FAISS

vectorstore = FAISS.from_documents(
    documents = split_docs,
    embedding = embeddings
)

print("chunk 개수 :", len(split_docs))
print("FAISS 벡터 개수 :", vectorstore.index.ntotal)
```

```python
results = vectorstore.similarity_search("코히어가 데이터 출처 탐색기를 공개한 이유는?")

print("검색 결과 개수 :", len(results))

print("\n===== 첫 번째 검색 결과 =====")
print(results[0].page_content)

print("\n===== metadata =====")
print(results[0].metadata)
```

```python
retriever = vectorstore.as_retriever(
    search_kwargs = {"k": 3}
)

results = retriever.invoke(
    "코히어가 데이터 출처 탐색기를 공개한 이유는?"
)

print("검색 결과 개수 :", len(results))

for i, doc in enumerate(results):
    print(f"\n===== 결과 {i} =====")
    print(doc.page_content[:100])
    print("metadata:", doc.metadata)
```

```python
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 3}
)

question = "코히어가 데이터 출처 탐색기를 공개한 이유는?"

results = retriever.invoke(question)

print("가져온 문서 수:", len(results))
```

```python
context = "\n\n".join(
    doc.page_content
    for doc in results
)

print(context)
```

```python
prompt = f"""
다음 문서를 참고해서 질문에 답해줘.

[문서]
{context}

[질문]
{question}
"""

response = llm.invoke(prompt)

print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""
다음 문서를 참고해서 질문에 답해줘.

[문서]
{context}

[질문]
{question}
""")


chain = prompt | llm
```

```python
response = chain.invoke({
    "context": context,
    "question": question
})

print(response.content)
```

```python
from langchain_core.runnables import RunnablePassthrough

def format_docs(docs):
    return "\n\n".join(
        f"[출처: PDF {doc.metadata.get('page_label', '?')}페이지]\n"
        f"{doc.page_content}"
        for doc in docs
    )


rag_chain = (
    {
        "context": retriever | format_docs,
        "question": RunnablePassthrough()
    }
    | prompt
    | llm
)
```

```python
question = "코히어가 데이터 출처 탐색기를 공개한 이유는?"

response = rag_chain.invoke(question)

print(response.content)

for i, doc in enumerate(results):
    print(f"\n===== 결과 {i} =====")
    print("페이지:", doc.metadata.get("page_label"))
    print("내용:", doc.page_content[:50])
```

```python
question = "코히어가 데이터 출처 탐색기를 공개한 이유는?"

similarity_retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 3}
)

mmr_retriever = vectorstore.as_retriever(
    search_type="mmr",
    search_kwargs={"k": 3}
)

similarity_results = similarity_retriever.invoke(question)
mmr_results = mmr_retriever.invoke(question)

print("===== similarity =====")

for i, doc in enumerate(similarity_results):
    print(f"\n--- {i} ---")
    print(doc.page_content[:200])
    
print("\n===== MMR =====")

for i, doc in enumerate(mmr_results):
    print(f"\n--- {i} ---")
    print(doc.page_content[:200])
```

```python
results = vectorstore.similarity_search_with_score(
    "코히어가 데이터 출처 탐색기를 공개한 이유는?",
    k=3
)

for i, (doc, score) in enumerate(results):
    print(f"\n===== 결과 {i} =====")
    print("점수:", score)
    print("내용:", doc.page_content[:300])
```

# PDF Loader

| Loader                | 핵심 기능                       | 출력 단위              |
| --------------------- | --------------------------- | ------------------ |
| **PyPDFLoader**       | 기본 PDF 읽기                   | **페이지 단위**         |
| **Unstructured**      | 문서 구조(제목/본문/목록 등) 분석        | **요소(Element) 단위** |
| **PDFPlumber**        | 표/레이아웃 처리                   | **페이지 단위**         |
| **PyMuPDF**           | 빠른 PDF 처리 + PDF metadata 추출 | **페이지 단위**         |
| **PyPDFium2**         | PDF 텍스트 추출                  | **페이지 단위**         |
| **PDFMiner**          | PDF 텍스트를 세밀하게 추출            | **전체 문서 단위**       |
| **PDFMinerPDFasHTML** | PDF → HTML 변환 → 레이아웃 정보 활용  | **전체 문서 단위**       |
| **PyPDFDirectory**    | 폴더 안의 여러 PDF를 한꺼번에 로드       | **PDF별 페이지 단위**    |
| **OCR**               | 이미지 PDF의 글자를 텍스트로 인식        | **OCR 결과 단위**      |
