```python
from langchain_openai import ChatOpenAI
```

```python
# 쳇모델 객체 생성
llm = ChatOpenAI(
    temperature=0.1, # 창의성 (0.0 ~ 2.0)
    model = "gpt-4o-mini", # 모델명 
)
```

```python
# 질문
question = "천안 KTX역 위치가 어디인가요?"
```

```python
# 응답, invoke 호출 메서드
response = llm.invoke(question)
```

### `logprob` 로그 프로버블리티 활성화
- 모델의 답변 확신도(Confidence level) : 0에 가까울수록 정답에 가까운 확신을 가진다.

```python
# Obj 생성
lim_with_logprob = ChatOpenAI(
    temperature=0.1,
    max_completion_tokens= 2048,
    model = "gpt-4o-mini"
).bind(logprobs=True)
```

```python
question2 = "인도의 수도는 어디인가요?"

response2 = lim_with_logprob.invoke(question2)
```

```python
# 스트림 방식으로 질의
answer = llm.stream("천안아산역 맛집 10곳과 주소를 알려주세요!")
```

```python
for token in answer:
    print(token.content, end="", flush=True)
```

```python
answer2 = llm.stream("점심 메뉴 5개 추천해주세요!")

final_answer = ""
for token in answer2:
    final_answer += token.content
```
