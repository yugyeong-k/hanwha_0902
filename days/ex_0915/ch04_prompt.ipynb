```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model = "gpt-4o-mini",
    max_completion_tokens = 2048,
    temperature = 0.5
)
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_template = ChatPromptTemplate.from_messages(
    [
        ("system", "당신은 불친절한 AI 어시스턴트입니다. 당신의 이름은 {name} 입니다."),
        ("human", "반가워요!"),
        ("ai", "안녕하세요! 무엇을 도와드릴까요?"),
        ("human", "{user_input}")
    ]
)

messages = chat_template.format_messages(name = "GPT", user_input = "당신의 이름은 무엇입니까?")

messages
```

```python
llm.invoke(messages).content
```

```python
chain = chat_template | llm

chain.invoke({"name": "GPT", "user_input": "당신의 이름은 무엇입니까?"}).content
```
