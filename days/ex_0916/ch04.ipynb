```python
import os

import langchain
import langchain_core
import langchain_openai

from dotenv import load_dotenv


# .env 로드
load_dotenv()


# API Key 확인
print(
    "[OK] OpenAI API 키 설정됨"
    if os.getenv("OPENAI_API_KEY")
    else "[ERROR] OpenAI API 키 미설정"
)

# LangChain Key
print(
    "[OK] LangChain 키 설정됨"
    if os.getenv("LANGSMITH_API_KEY")
    else "[ERROR] LangChain 키 미설정"
)

# LangChain 버전 확인
print(f"LangChain: {langchain.__version__}")
print(f"LangChain Core: {langchain_core.__version__}")
print(f"LangChain OpenAI: {langchain_openai.__version__}")
```

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model("gpt-4o-mini", temperature=0.2, max_tokens=2048)
```

```python
from pydantic import BaseModel, Field
from langchain_core.output_parsers import JsonOutputParser
from langchain_core.prompts import ChatPromptTemplate
```

```python
class Topic(BaseModel):
    description : str = Field(description = "주제에 대한 간결한 설명")
    hashtag     : str = Field(description = "해시태그 형식의 키워드, 2개 이상")

question = "AI 모델 중 제일 똑똑한 모델을 알려주세요."

parser = JsonOutputParser(pydantic_object=Topic)

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "당신은 분석 능력이 뛰어난 AI 어시턴트 입니다. 질문에 간결하게 답변하세요."),
        ("user", "#Fomat: {format_instructions}\n\n#Question: {question}"),
    ]
)

prompt = prompt.partial(format_instructions = parser.get_format_instructions())

chain = prompt | llm | parser

chain.invoke({"question": question})
```
