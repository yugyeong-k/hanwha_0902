```python
import os

import langchain
import langchain_core
import langchain_openai

from dotenv import load_dotenv


# .env 로드
load_dotenv()


# API Key 확인
print(
    "[OK] OpenAI API 키 설정됨"
    if os.getenv("OPENAI_API_KEY")
    else "[ERROR] OpenAI API 키 미설정"
)

# LangChain Key
print(
    "[OK] LangChain 키 설정됨"
    if os.getenv("LANGSMITH_API_KEY")
    else "[ERROR] LangChain 키 미설정"
)

# LangChain 버전 확인
print(f"LangChain: {langchain.__version__}")
print(f"LangChain Core: {langchain_core.__version__}")
print(f"LangChain OpenAI: {langchain_openai.__version__}")
```

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model("gpt-4o-mini", temperature=0.2, max_tokens=2048)
```

```python
from pydantic import BaseModel, Field
from langchain_core.prompts import PromptTemplate
```

```python
class Response(BaseModel):
    answer: str = Field(description="질문에 대한 답변")
    source: str = Field(description="답변의 출처 또는 웹사이트 주소")

response_prompt = PromptTemplate(
    template="사용자의 질문에 최대한 답변하세요.\n{question}",
    input_variables=["question"],
)

structured_llm = llm.with_structured_output(Response)

chain = response_prompt | structured_llm

result = chain.invoke({"question": "대한민국의 수도는 어디인가요?"})

print("전체 결과:", result)
print("답변:", result.answer)
print("출처:", result.source)
```

```python
stream_chain = response_prompt | llm

for chunk in stream_chain.stream({"question": "AI중에 누가 제일 똑똑한가요?"}):
    print(chunk.content, end="", flush=True)
```
